### 📝 Respostas do Exercício 1

#### 1. Que tipos de dado são aceitos dentro de `metadata`? O que acontece ao colocar uma lista ou dicionário aninhado?
* **Tipos aceitos:** No objeto `Document` do LangChain, o campo `metadata` é um dicionário Python comum (`dict`). Portanto, ele aceita **qualquer tipo de dado nativo do Python**: textos (`str`), inteiros (`int`), decimais (`float`), booleanos (`bool`), listas (`list`), tuplas (`tuple`) e dicionários aninhados (`dict`).
* **Resultado do teste:** Ao colocar uma lista (`"tags": ["vetores", "nlp"]`) ou um dicionário aninhado (`"config": {"dimensao": 384}`), o objeto é criado com total sucesso sem gerar nenhum erro.
* *Nota prática:* Embora o LangChain aceite qualquer tipo em memória, alguns bancos de dados vetoriais (como ChromaDB ou Pinecone) exigem apenas tipos primitivos (`str`, `int`, `float`, `bool`) caso você use filtros nos metadados.

#### 2. O que acontece se você criar um `Document` sem passar `metadata`?
* Se instanciarmos `Document(page_content="...")` omitindo o parâmetro `metadata`, **o código não quebra nem gera erro**.
* O LangChain automaticamente define o valor padrão de `metadata` como um dicionário vazio: `{}`.
* Isso garante que `doc.metadata` sempre retorne um `dict` válido, permitindo que seu código faça acessos e consultas (`doc.metadata.get(...)`) com segurança.

````python
import json
from langchain_core.documents import Document

# 1. Definição do chunk real seguindo rigorosamente o Schema do Notion + 3 Campos Próprios
chunk_real_metadata = {
    # Campos Obrigatórios da Tabela
    "fonte": "documento_01.md",
    "documento_id": "documento_01",
    "chunk_index": 12,
    "estrategia": "recursive_hierarchical",  # Teste 9 da Aula 04
    "chunk_size": 600,
    "chunk_overlap": 60,
    "n_caracteres": 534,

    # 3 Campos Próprios Adicionais
    "pagina": 3,
    "secao": "Metodologia de Divisão Textual",
    "n_tokens": 118
}

chunk_real = Document(
    page_content="O RecursiveCharacterTextSplitter avalia recursivamente uma lista de separadores hierárquicos para preservar a estrutura semântica original do texto antes de realizar cortes arbitrários.",
    metadata=chunk_real_metadata
)

# 2. Exibição do JSON formatado
print("✅ Exemplo preenchido em JSON do Chunk Real:\n")
print(json.dumps({
    "page_content": chunk_real.page_content,
    "metadata": chunk_real.metadata
}, ensure_ascii=False, indent=2))
````

### 📋 Exercício 2 - Schema de Metadados Projetado

#### 1. Tabela com o Schema Final

| Campo | Tipo | Origem | Descrição e Função no RAG |
| :--- | :---: | :---: | :--- |
| `fonte` | `str` | Obrigatório | Nome do arquivo `.md` de origem (ex.: `documento_01.md`). |
| `documento_id` | `str` | Obrigatório | Identificador único do documento (ex.: `documento_01`). |
| `chunk_index` | `int` | Obrigatório | Posição sequencial do chunk dentro do documento original. |
| `estrategia` | `str` | Obrigatório | Estratégia da Aula 04 utilizada (ex.: `recursive_hierarchical`). |
| `chunk_size` | `int` / `str` | Obrigatório | Configuração de tamanho máximo utilizada na divisão. |
| `chunk_overlap` | `int` | Obrigatório | Configuração de sobreposição de caracteres utilizada. |
| `n_caracteres` | `int` | Obrigatório | Tamanho real em caracteres do texto do chunk. |
| **`pagina`** | `int` | **Próprio 1** | Número da página física do documento PDF original. |
| **`secao`** | `str` | **Próprio 2** | Nome da seção ou título Markdown a que o chunk pertence. |
| **`n_tokens`** | `int` | **Próprio 3** | Contagem exata de tokens do chunk no modelo de linguagem. |

---

#### 2. Justificativa de Cada Campo Próprio (Perguntas que permite responder)

* **`pagina` (int):**
  * *Que pergunta responde?* "Em qual página física do artigo original este trecho foi extraído?"
  * *Utilidade:* Permite ao sistema RAG criar citações precisas para o usuário humano (ex.: *"Fonte: Artigo X, página 3"*), facilitando a auditoria e a verificação da informação.

* **`secao` (str):**
  * *Que pergunta responde?* "A qual tópico, capítulo ou contexto temático este chunk pertence?"
  * *Utilidade:* Permite aplicar pré-filtros na busca vetorial (ex.: pesquisar apenas dentro da seção *"Metodologia"* ou *"Resultados"*), aumentando a precisão da recuperação.

* **`n_tokens` (int):**
  * *Que pergunta responde?* "Quantos tokens este chunk consumirá na janela de contexto do LLM?"
  * *Utilidade:* Evita o estouro da janela de contexto no momento de montar o prompt de resposta do RAG, selecionando dinamicamente o número ideal de chunks recuperados.

#### Justificativa dos Campos Adicionais
* **`pagina` (int):** Permite responder *"Em qual página do artigo original este trecho está?"*, viabilizando citações exatas da fonte.

#### Perguntas Adicionais
* **Citação da fonte:** Para citar a fonte de forma precisa na resposta final, usam-se os campos **`pagina`** e **`secao`** (junto com **`fonte`**), permitindo ao modelo citar: *"Segundo documento_01.md (Seção 'Metodologia', pág. 3)..."*.
* **Utilidade do `chunk_index`:** Garante a ordem sequencial dos trechos. Se um chunk for recuperado com uma frase ou raciocínio cortado pela metade, o sistema pode recuperar diretamente os vizinhos (`chunk_index - 1` e `chunk_index + 1`) sem precisar de uma nova busca vetorial (*Context Expansion*).
