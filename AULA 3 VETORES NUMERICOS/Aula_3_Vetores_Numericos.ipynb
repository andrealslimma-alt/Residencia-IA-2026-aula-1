```python
# 1. INSTALA A BIBLIOTECA DE EMBEDDINGS GRATUITA
!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer, util

# 2. CARREGA O MODELO LEVE DE EMBEDDINGS (Roda direto na memória do Colab)
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. BASE DE DADOS (JSON)
documents = [
    {
        "titulo": "O caso Twitter/X: algoritmo, espaço público e moderação de conteúdo",
        "autores": ["Equipe de Pesquisa"],
        "ano": 2025,
        "resumo": "Este artigo investiga a dinâmica das redes sociais na formação do debate público, com ênfase no Twitter (atualmente denominado X) e na influência dos algoritmos na regulação do conteúdo. Por meio de uma abordagem interdisciplinar...",
        "palavras_chave": ["Algoritmos", "Redes Sociais", "Moderação de Conteúdo", "Twitter/X"],
        "categoria_tema": "Tecnologia, Algoritmos e Sociedade",
        "arquivo_origem": "twitter_algoritmo.md"
    },
    {
        "titulo": "Entre o algoritmo e o Juramento de Hipócrates",
        "autores": ["Equipe de Pesquisa"],
        "ano": 2022,
        "resumo": "O avanço da inteligência artificial tem transformado profundamente a prática médica. De sistemas de apoio à decisão clínica a algoritmos de triagem e diagnóstico, a inteligência artificial tem demonstrado potencial para diagnósticos precoces, terap...",
        "palavras_chave": ["Bioética", "Inteligência Artificial", "Ética Médica", "Saúde Digital"],
        "categoria_tema": "Bioética e Saúde Digital",
        "arquivo_origem": "bioetica_e_ia (1).md"
    },
    {
        "titulo": "Escrita acadêmica ética, responsável e Inteligência Artificial",
        "autores": ["Equipe de Pesquisa"],
        "ano": 2025,
        "resumo": "I Programa Pós-Graduação em Ciência Política, Universidade Federal do Paraná, Curitiba, PR, Brasil.",
        "palavras_chave": ["Escrita Acadêmica", "Inteligência Artificial", "Metodologia", "Integridade Científica"],
        "categoria_tema": "Educação, Pesquisa e IA Generativa",
        "arquivo_origem": "escrita_academica_ia.md"
    }
]

# 4. FORMATAÇÃO DO TEXTO
def doc_to_text(doc: dict) -> str:
    palavras_chave = ", ".join(doc.get("palavras_chave", []))
    return (
        f"Título: {doc.get('titulo')}. "
        f"Categoria: {doc.get('categoria_tema')}. "
        f"Palavras-chave: {palavras_chave}. "
        f"Resumo: {doc.get('resumo')}"
    )

# 5. EXECUÇÃO DA BUSCA SEMÂNTICA
doc_texts = [doc_to_text(d) for d in documents]
query = "redes sociais e algoritmos de moderação"

# Gera os vetores e calcula a similaridade
query_vec = model.encode(query, convert_to_tensor=True)
doc_vecs = model.encode(doc_texts, convert_to_tensor=True)
cosine_scores = util.cos_sim(query_vec, doc_vecs)[0]

# Organiza os resultados
resultados = []
for doc, score in zip(documents, cosine_scores):
    resultados.append({
        "titulo": doc["titulo"],
        "categoria_tema": doc["categoria_tema"],
        "arquivo_origem": doc["arquivo_origem"],
        "similarity": float(score)
    })

resultados.sort(key=lambda x: x["similarity"], reverse=True)

# 6. IMPRESSÃO DOS RESULTADOS
print("=== RESULTADOS DA BUSCA SEMÂNTICA ===")
print(f"Consulta: '{query}'\n")

for i, res in enumerate(resultados, start=1):
    print(f"{i}. Título: {res['titulo']}")
    print(f"   Categoria: {res['categoria_tema']}")
    print(f"   Arquivo: {res['arquivo_origem']}")
    print(f"   Similaridade: {res['similarity']:.4f}\n")
```
